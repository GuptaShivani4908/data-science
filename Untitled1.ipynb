```python
# assigning values to variables
x = 10       # integer
name = "Riya" # string
pi = 3.14    # float
is_hot = True # boolean
```

```python
a = 10
b = -25
print(a, type(a))  # 10 <class 'int'>
print(b, type(b))  # -25 <class 'int'>
```

```python
pi = 3.1415
g = -9.8
print(pi, type(pi))  # 3.1415 <class 'float'>
```

```python
z = 2 + 3j
print(z, type(z))   # (2+3j) <class 'complex'>
print(z.real)       # 2.0
print(z.imag)       # 3.0
```

```python
x = True
y = False
print(x, type(x))   # True <class 'bool'>
print(10 > 5)       # True
print(10 < 5)       # False
```

```python
name = "Python"
sentence = 'I love coding'
multiline = """This is 
a multiline string"""
print(name, type(name))  # Python <class 'str'>
```

```python
fruits = ["apple", "banana", "cherry"]
fruits[1] = "mango"   # modifying list
print(fruits, type(fruits))  # ['apple', 'mango', 'cherry']
```

```python
colors = ("red", "green", "blue")
print(colors, type(colors))  # ('red', 'green', 'blue') <class 'tuple'>
```

```python
r = range(5)   # 0,1,2,3,4
print(list(r)) # [0, 1, 2, 3, 4]
```

```python
student = {"name": "Riya", "age": 20, "course": "CS"}
print(student["name"])   # Riya
print(student, type(student))  # {'name': 'Riya', 'age': 20, 'course': 'CS'}
```

```python
nums = {1, 2, 3, 2, 1}
print(nums, type(nums))  # {1, 2, 3}
```

```python
fs = frozenset([1, 2, 3, 3])
print(fs, type(fs))  # frozenset({1, 2, 3})
```

```python
b = b"hello"
print(b, type(b))  # b'hello' <class 'bytes'>
```

```python
ba = bytearray([65, 66, 67])
ba[0] = 68
print(ba)          # bytearray(b'DBC')
```

```python
m = memoryview(bytes([1, 2, 3, 4]))
print(m[0])  # 1
```

```python
x = None
print(x, type(x))  # None <class 'NoneType'>
```
