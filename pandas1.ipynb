```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 78, 92],
    'Science': [90, 82, 89],
    'English': [88, 85, 94]
}

df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
#using melt
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
```

```python
df
```

```python
df.groupby("Department")["Salary"].mean() #average salary dikhne ke liye
```

```python
type(df.groupby("Department")["Salary"].mean())
```

```python
df.groupby("Team")["Salary"].sum()
```

```python
df.groupby("Team")["Salary"].max()
```

```python
df.groupby("Team")["Salary"].min()
```

```python
df.groupby("Team")["Salary"].count()
```

```python
# .agg and aggregate
```

```python
df.groupby("Team")["Salary"].agg(["mean", "max", "min","count"])
```

```python
df.groupby("Team")["Salary"].agg(
    avg_score="mean",
    high_score="max"
)
```

```python
df["Team Avg"] = df.groupby("Team")["Salary"].transform("mean")
```

```python
df
```

```python
df.groupby("Team").filter(lambda x: x["Salary"].mean() > 85)
```

```python
# Merging & Joining Data
```

```python
employees = pd.DataFrame({"EmpID": [1,2,3], "Name" :["Alice","Bob","shiv"], "DeptID": [10,20,30]})
```

```python
employees
```

```python
departments = pd.DataFrame({ "DeptID": [10,20,40], "DeptName":["HR","Engineering","Marketing"]})
```

```python
departments
```

```python
pd.merge(employees,departments, on = "DeptID") #inner join
```

```python
pd.merge(employees,departments, on = "DeptID", how ="left") #left join
```

```python
pd.merge(employees,departments, on = "DeptID", how ="right") #right join
```

```python
pd.merge(employees,departments, on = "DeptID", how ="outer") #outer join
```
