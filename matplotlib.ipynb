```python
import numpy as np 
years =[2001, 2002, 2003, 2004, 2005, 2006, 2007, 2008, 2009, 2010]
kohli= [0,500,800,1100,1300,1500,1800,1900,2100,900]
Rohit =[0,100,200,500,1500,1200,2000,500,2200,1000]
Sehwag =[0,300,800,1200,1500,1700,1600,1400,1000,0]

x = np.arange(len(years))
width=0.25

plt.bar(x - width , kohli, width=width, label="kohli")
plt.bar(x, Rohit, width=width, label="Rohit")
plt.bar(x + width , Sehwag, width=width , label="Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Runs Comparison")
plt.xticks(x,years)
plt.legend()
plt.tight.layout()
plt.show()
```

```python
years =[2001, 2002, 2003, 2004, 2005, 2006, 2007, 2008, 2009, 2010]

import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))  # index positions
width = 0.25



plt.bar(x - width, sachin, width=width, label="Sachin")
plt.bar(x, sehwag, width=width, label="Sehwag")
plt.bar(x + width, kohli, width=width, label="Kohli")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Run Comparison")
plt.xticks(x, years)  # Show actual year(2003) instead of 0,1,2,...
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs, color="lightgreen")

# Add labels on top of bars
for i in range(len(players)):
    plt.text(i, runs[i] + 30, str(runs[i]), ha='center')

plt.ylabel("Runs")
plt.title("Runs Scored by Players")
plt.tight_layout()
plt.show()
```

```python
# Pie Charts in Matplotlib
```

```python
plt.style.use("ggplot")
```

```python
labels=["sachin","sehwag","kohli","yuvraj"]
runs =[18000,8000,12000,3000]

plt.title("career runs")
plt.pie(runs, labels=labels)
plt.show()
```

```python
labels=["sachin","sehwag","kohli","yuvraj"]
runs =[18000,8000,12000,3000]

colors=['lightpink','skyblue','lightgreen','#654321']
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

```python
plt.pie(
    runs, 
    labels=labels,
    colors=colors, 
    wedgeprops={'edgecolor':'green', 'linewidth':2,'linestyle':'--', 'fill':False}
)
plt.show()
```

```python
plt.pie(
    runs, 
    labels=labels,
    colors=colors, 
    wedgeprops={'edgecolor':'green', 'linewidth':2,'linestyle':'--', 'fill':True}
)
plt.show()
```

```python
explode =[0.1,0. ,0 ,0]

plt.pie(runs, labels=labels, colors = colors, explode=explode, shadow=True, startangle=120, autopct='%1.1f%%')
plt.title("exploded")
plt.show()
```
