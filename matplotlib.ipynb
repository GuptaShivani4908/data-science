```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3],[4,5,6])
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python
plt.plot(years,runs)
```

```python
plt.xlabel("years")
```

```python
plt.ylabel("runs scored")
```

```python
plt.title("sachin tendulkars yearly runs")
```

```python
plt.show()
```

```python
plt.plot(years, runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
plt.show()
```

```python
kohli= [0,0,500,800,1100,1300,1500,1800,1900,2100]
sehwag =[0,300,800,1200,1500,1700,1600,1400,1000,0]
plt.plot(years,kohli, label="virat kohli")
plt.plot(years,sehwag,label="virende sehwag")

plt.xlabel("year")
plt.ylabel("runs scored")
plt.title("performance comparison")
plt.legend()
plt.show()
```

```python
kohli= [0,0,500,800,1100,1300,1500,1800,1900,2100]
sehwag =[0,300,800,1200,1500,1700,1600,1400,1000,0]
plt.plot(years,kohli)
plt.plot(years,sehwag)

plt.legend(["kholi","sehwag"])

plt.xlabel("year")
plt.ylabel("runs scored")
plt.title("performance comparison")

plt.show()
```

```python
plt.plot(years,kohli,'ro--', label="kohli") #r = red color , o =circlevformat
plt.plot(years,sehwag,'g^:',label="sehwag")#g= green , ^= triangle
plt.legend()
```

```python
plt.plot(years,kohli,'-g', label="kohli") #r = red color , o =circlevformat
plt.plot(years,sehwag,'b-.',label="sehwag")#g= green , ^= triangle
plt.tight_layout()
plt.legend()
```
